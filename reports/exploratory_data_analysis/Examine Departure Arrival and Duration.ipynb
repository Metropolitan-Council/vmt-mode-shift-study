```python
df = pd.read_csv(cleaned_file)
df.head()
```

```python
# clean up departure and arrival timestamps
df['depart_time'] = df['depart_time'].apply(pd.Timestamp)
df['arrive_time'] = df['arrive_time'].apply(pd.Timestamp)
```

```python
df['duration_check'] = (df.arrive_time - df.depart_time).apply(pd.Timedelta.total_seconds) / 60.
```

```python
df['duration_check'] = df['duration_check'].apply(lambda x : x+1440 if x<0 else x)
```

```python
df.head()
```

```python
sns.scatterplot(df, x='duration', y='duration_check')
```

```python
len(df)
```

```python
df2 = df[((df.duration_check - df.duration)>1) | ((df.duration_check - df.duration)<-1)]
```

```python
df2['depart_time'] = df2['depart_time'].apply(pd.datetime.time)
df2['arrive_time'] = df2['arrive_time'].apply(pd.datetime.time)
```
