```python
import pandas as pd
import geopandas as gpd
import keyring
import os
from tqdm import tqdm
```

```python
# get base path for data
data_dir = keyring.get_password("msp", "vmt_reduction_dir")
```

```python
files = os.listdir(data_dir + "/Data_Processed/car_congestion")
```

```python
dfs = []
for file in files:
    print(file)
    curr = gpd.read_file(data_dir + "/Data_Processed/car_congestion/" + file, ignore_geometry=True)
    dfs.append(curr)
```

```python
len(dfs)
```

```python
df = pd.concat(dfs, keys=[temp[4:-5] for temp in files])
```

```python
df = df.reset_index().set_index(["level_0", "trip_id"])
```

```python
df = df.drop("level_1", axis=1)
```

```python
df
```

```python
df.to_parquet(data_dir + r"/Data_processed/geodata/car_congestion.parquet")
```
